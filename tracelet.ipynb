## Probabilistic and modular ReACT

The goal of this notebook is to outline a probabilistic variant of ReACT prompting for agentic workflows. 

In a nutshell, ReACT and many variants can be expressed as the following program:
```python
context = query 

while True: 
    thought, action ~ LLM(. | context)
    if action == "stop": 
        break 
    observation = exec(action)
    context += (thought + action + observation)  
```
which results in an output *trajectory* $\tau_{\text{ReACT}}$:
$$\tau_{\text{ReACT}} = \bigg\{ (t_{1}, a_{1}, o_{1}), (t_{2}, a_{2}, o_{2}), ..., (t_{n}, a_{n}, o_{n})  \bigg\}$$
where $t_{j}, a_{j}, o_{j}$ refer to *thoughts, actions* and *observations* at step $j$.  

Below shows a concrete example using [**smolagents**](https://github.com/huggingface/smolagents) for the query 
> How many seconds would it take for a leopard at full speed to run through Pont des Arts?

````python
model = OpenAIModel(
    model_id="gpt-4o"
)
agent = CodeAgent(
    tools=[WebSearchTool()], 
    model=model, 
    stream_outputs=False
)
````

````python
agent.run("How many seconds would it take for a snow leopard at full speed to run through Pont des Arts?")
````

where the output of both the actions and execution output are logged. For example, in the first step, the action proposed is to do a web search to find the length of *Pont des Arts bridge*: 

> pont_des_arts_length = web_search("length of Pont des Arts bridge")                                              
> print(pont_des_arts_length)

which is following by another search for the speed of a snow leopard and a call to generate code to compare the two values (*note that the output will change each time this is run, so the trajectory might not match this sequence exactly*). 

The final answer based on this computation is then reported:  
> Final answer: XXXXXXXX 

Standard ReACT and its many variations have the <span style="color:red">**following shortcomings**</span>: 

- **Context lenght quickly becomes a bottleneck** given that *thoughts, actions, observations* often get naively concatenated into the running context for the LLM, which is seen in the line below. While the example above is simple, it requires 10-15k LLM tokens, mostly due to the length of the search results that get continuously added to the running context at each step. 
```python
context += (thought + action + observation)
``` 
- **No explicit uncertainty**. Given that ReACT operates entirely in the text domain and returns a single answer, no explicit account of model uncertainty is provided. This makes it hard to reason about the robustness of the analysis. For example, running the query above results in a wide range of different numbers; ideally one would want to compute an expectation over these different values. 
- **Modularity is removed**. For difficult problems, often the model will make multiple attempts to solve the same sub-problem. In the unrolled trajectory, this looks like the following: $$\tau_{\text{ReACT}} = \bigg\{ \underbrace{(t_{1}, a_{1}, o_{1}), (t_{2}, a_{2}, o_{2}), ...}_{\text{same action } a, \text{many failures/revisions}}, (t_{n}, a_{n}, o_{n})  \bigg\}$$where many subsequent steps focus on the same underlying problem with many failed attempts or revised inputs to the tools. This lack of modularity seem foundational to many of the problems above: it makes it difficult to reason about the structure of the underlying reason process, manage context more effectively and diagnose or optimize potential pain points in the model (e.g., high cost sub-processes, low precision actions, etc.)

The <span style="color:red">**goal of this project**</span> is to develop an alternative variant of ReACT that: 
- Facilitate correct and formally precise probabilistic reasoning; 
- Improves the time and token efficiency of naive ReACT workflows and closely related variants;
- Enables detailed diagnosis of model behavior and the potential for algorithmically precise, agentic workflow optimization.  

## Tracelets 

Our approach is based on the notion of a **tracelet**, defined as a contiguous sequence of tuples in a Reason+Action trace that involve a shared action $a$ and goal. Using the example ReACT trace discussed above: 
$$\tau_{\text{ReACT}} = \bigg\{ \underbrace{(t_{1}, a_{1}, o_{1}), (t_{2}, a_{2}, o_{2}), ...}_{\text{tracelet with action } a_{1}=a_{2}, \text{shared goal}}, (t_{n}, a_{n}, o_{n})  \bigg\}$$
the failed actions noted above might be an example of a tracelet. For example, using our running example suppose that action $a_{1}$ is:

```python
pont_des_arts_length = web_search("length of Pont des Arts bridge")
print(pont_des_arts_length)
```

and the second action $a_{2}$ is: 

```python
pont_des_arts_length = web_search("find me the length of Pont des Arts bridge")
print(pont_des_arts_length)
```
Both involve the same underlying action (i.e., make a web search) with the same goal (i.e., determine the lenght of Pont des Arts bridge) just using different arguments.

Building on our past work on [**Probabilistic Programs of Thought (PPoT)**](https://arxiv.org/pdf/2604.17290), we can explicitly represent such as a tracelet declaratively as the following probabilistic program: 
```python
### sample from model 
prefix = "generate an argument for: pont_des_arts_length = web_search("
C ~ categorical(LLM(. | context, prefix))

## perform web search 
pont_des_arts_length = web_search(C)

return pont_des_arts_length 
```
where the LLM generation decision is modeled formally as a random variable. Operationally, one can query the model upfront for multiple samples as in the below *concrete* sampling implementation: 
```python
prefix = "generate an argument for: pont_des_arts_length = web_search("
num_samples ~ N
S ~ Sample(
    LLM, num_samples, context, prefix, action
)
return S
```
Where the sub-routine **Sample** can take many different concrete forms, each involving different decisions about how context is shared across samples (e.g., sampling without replacement via self-refine, sampling by repeated i.i.d calls to the model). 

#### ReACT with tracelets 

Below is a modified version of ReACT that adds a **Tracelet** subroutine that takes an *agent_spec* and produces upfront the kinds of tracelet programs described above. 
```python
context = query 

while True: 
    thought, action_spec ~ LLM(. | context)
    if action_spec == "stop": 
        break 
    trace = Tracelet(thought, action_spec, context)
    context += trace
```
resulting in a trajectory that is effectively a sequence of **Tracelet Programs**.
